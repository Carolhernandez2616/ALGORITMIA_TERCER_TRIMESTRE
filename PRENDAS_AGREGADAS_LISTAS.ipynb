```python
#LISTAS
#Que es una lista
#Es una estructura ordenada y mutable = puede agregar ,quitar y modificar elementos
#Puede contener elementos int, float, bool, dict.....
#======================================
#Una tienda desea guardar la lsta de nombres de ss clientes registrados
#Para promociones. El sistema debe hacer:
#1. Agregar nuevos clientes append()
#2. Recorrer la lista y mostrar los clientes for
#3. Modificar un nombre en caso de error
#4. Borrar un cliente

#======================================
#SOLUCION
#======================================

#1.Agregar nuevos clientes append()

def agregar_prendas(lista_prendas, nombre):
    if isinstance(nombre, str) and 2 <= len(nombre) <= 50:
        lista_prendas.append(nombre)
        print(f"\nPRENDA AGREGADA: {nombre}")

    else:
        print("\nNOMBRE DE LA PRENDA INVALIDO DEBE TENER ENTRE 2 A 50 CARACTERES")


#2. Recorrer la lista y mostrar los clientes for
def mostrar_prendas(lista_prendas):
    for prendas in lista_prendas:
        print(f"-> {prendas}")

#3.Modificar un nombre en caso de error
def  modificar_prendas(lista_prendas, indice , nuevo_nombre):
    if isinstance(nuevo_nombre, str) and 2 <= len(nuevo_nombre) <= 50:
        print("\nNOMBRE DE  PRENDA INVALIDO DEBE TENER ENTRE 2 A 50 CARACTERES")
        return #Termina la funcion
    if 0 <= indice < len(lista_prendas):
        nombre_original = lista_prendas[indice]
        lista_prendas[indice] = nuevo_nombre.title()
        print(f"\nLA PRENDA {nombre_original} FUE REEMPLAZADA POR: {nuevo_nombre.title()}")
    else:
        print("\nNO SE ENCONTRO EL DATO, NUMERO DE LA LISTA ERRONEO")

#4Borrar  un cliente
def borrar_prendas(lista_prendas, indice):
    if 0 <= indice < len(lista_prendas):
        borrar = lista_prendas.pop(indice)
        print(f"\nLA PRENDA {borrar}, SE BORRO DE LA LISTA DE PRENDA")
    else:
        print("\nNO SE ENCONTRO LA PRENDA, PRENDA NO BORRADO DE LA LISTA.")

def main():

#Simulacion empresarial
   prendas =  ["CAMISAS", "PANTALONES", "ZAPATOS", "MEDIAS", "INTERIORES", "CHAQUETAS", "VESTIDOS","RUANAS"]
   print("PRENDAS ACTUALES")
   mostrar_prendas(prendas)
   #Agregar nuevos clientes append()
   agregar_prendas(prendas, "JOYAS")
   mostrar_prendas(prendas)
   #Modificar un nombre en caso de error
   modificar_prendas(prendas,1,"CAMISA")
   mostrar_prendas(prendas)
   #Borrar un cliente
   borrar_prendas(prendas, 3)






if __name__ == "__main__":
    main()
```
